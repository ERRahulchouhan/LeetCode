```python
d = {'gfg': [1, 2, 3], 'is': [1, 4], 'best': [4, 2]}

f = {}
for key,i in d.items():
    for k in i:
        if k in f:
            f[k].append(key)
        else:
            f[k] = []
            f[k].append(key)

print(f)
```

```python
d = {'gfg': [1, 2, 3], 'is': [1, 4], 'best': [4, 2]}
from collections import defaultdict

f = defaultdict(list)
for key,i in d.items():
    for j in i:
        f[j].append(key)

print(f)
```

```python
# Python3 program to
# find n'th Lucas number

# recursive function
def lucas(n):
  
    # Base cases
    if n == 0:
        return 2;
    if n == 1:
        return 1;
  
    # recurrence relation
    return lucas(n - 1) + lucas(n - 2);
    
# Driver code to test above methods
n = 9;
print(lucas(n));

# This code is contributed by phasing17
```

```python
def nth_fibonacci(n):
    if n <= 1:
        return n

    return nth_fibonacci(n-1) + nth_fibonacci(n-2)

nth_fibonacci(5)
```
