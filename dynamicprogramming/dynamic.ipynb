```python
# A simple recursive CPP program to print
# first n Tribonacci numbers.

def printTribRec(n) :
    if (n == 0 or n == 1 or n == 2) :
        return 0
    elif (n == 3) :
        return 1
    else :
        return (printTribRec(n - 1) + 
                printTribRec(n - 2) +
                printTribRec(n - 3))
        

def printTrib(n) :
    for i in range(1, n) :
        print( printTribRec(i) , " ", end = "")
        

# Driver code
n = 10
printTrib(n)


# This code is contributed by Nikita Tiwari.
```

```python
def lucas(n):
    if n == 0:
        return 2
    if n == 1:
        return 1

    return lucas(n - 1) + lucas(n - 2)

lucas(4)
```

```python
def countWays(n):

    # Base cases
    if n == 0 or n == 1:
        return 1

    return countWays(n - 1) + countWays(n - 2)

n = 4
print(countWays(n))

for i in range(4):
    print(countWays(i))
```
