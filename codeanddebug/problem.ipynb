```python
def rev(n,l):
    if len(n) == 0:
        return l

    r = n.pop()
    l.append(r)

    return rev(n,l)


num = [1,2,3]
rev(num,[])
```

```python
def rev(n,l):
    if len(n) == 0:
        return l

    r = n[-1]
    l +=[r]
    n = n[:len(n)-1]

    return rev(n,l)


num = [1,2,3]
rev(num,[])
```

```python
def rev(num,left, right):
    if left >= right:
        return num

    num[left],num[right] = num[right], num[left]
    return rev(num,left+1 , right-1)

num = [1,2,3]
rev(num,0, len(num)-1)
```

```python
def palidorm():
    a = "ara"
    n = len(a)
    left = 0
    right = n-1

    while left <= right:
        if a[left] != a[right]:
            return False 
        left +=1
        right -=1
    return True
palidorm()
```

```python
def selec(num,newnum):
    if len(num) == 0:
        return newnum

    minimum = num[0]
    for i in range(1, len(num)):
        if num[i] < minimum:
            minimum = num[i]

    newnum.append(minimum)
    num.remove(minimum)

    return selec(num, newnum)

n  = [5,7,8,4,1,6,9,2]

selec(n,[])
```

```python
def selec(num,count):
    if count == len(num):
        return count

    if num[count] > num[count+1]:
        num[count] , num[count+1] =  num[count+1], num[count] 
    count +=1
    return selec(num, count)

n  = [5,7,8,4,1,6,9,2]

selec(n,0)
```

```python
def selec(num, count):

    if count == len(num) - 1:
        return

    if num[count] > num[count + 1]:
        num[count], num[count + 1] = num[count + 1], num[count]

    selec(num, count + 1)


n = [5, 7, 8, 4, 1, 6, 9, 2]

selec(n, 0)
print(n)
```

```python
def sort(num, count):
    if count == len(num) - 1:
        return
    selec(num, 0)
    sort(num, count + 1)


n = [5, 7, 8, 4, 1, 6, 9, 2]
sort(n, 0)

print(n)
```

# insertion sort

```python
num = [5, 7, 8, 4, 1, 6, 9, 2]
for i in range(1, len(num)):

    key = num[i]

    for j in range(i, 0, -1):

        if num[j-1] > key:
            num[j] = num[j-1]
        else:
            break

    num[j-1] = key


print(num)
```

```python
num = [5, 7, 8, 4, 1, 6, 9, 2]

for i in range(1, len(num)):
    j = i
    while j > 0:
        if num[j-1] > num[j]:
            num[j-1], num[j] = num[j], num[j-1]

        j -= 1

print(num)
```

```python
num = [5, 7, 8, 4, 1, 6, 9, 2]

for i in range(len(num)):
    key = num[i]
    j = i-1

    while j >= 0 and key < num[j]:
        num[j+1] = num[j]
        j-=1

    num[j+1] = key
print(num)
```
