```python
import yt_dlp

playlist_url = "https://www.youtube.com/watch?v=OtYEY2htIjM&list=PLhR2IpV1b2FwWwviBHRrR118YAaSlyhTU"

opts = {
    "extract_flat": "in_playlist",
    "quiet": False,
    "ignoreerrors": True,
}

with yt_dlp.YoutubeDL(opts) as ydl:
    info = ydl.extract_info(playlist_url, download=False)

if "entries" not in info:
    print("Playlist was not detected.")
    print(info)
else:
    entries = [v for v in info["entries"] if v and v.get("id")]

    with open("link.txt", "w", encoding="utf-8") as f:
        for video in entries:
            f.write(f"https://www.youtube.com/watch?v={video['id']}\n")

    print(f"Saved {len(entries)} links to link.txt")
```

```python
import yt_dlp
import os

os.makedirs("videos", exist_ok=True)

with open("link.txt", "r", encoding="utf-8") as f:
    urls = [x.strip() for x in f if x.strip()]

opts = {
    "format": "bestvideo[height<=720]+bestaudio/best[height<=720]/best",
    "merge_output_format": "mp4",
    "outtmpl": "videos/%(title)s.%(ext)s",
    "noplaylist": True,
}

with yt_dlp.YoutubeDL(opts) as ydl:
    ydl.download(urls)

print("Finished.")
```
